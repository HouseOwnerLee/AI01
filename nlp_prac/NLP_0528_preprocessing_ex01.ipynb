## 불용어

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from konlpy.tag import Okt
```

```python
stop_words_list = stopwords.words('english')
print('불용어 개수 :', len(stop_words_list))
print('불용어 10개 출력 :',stop_words_list[:10])
```

```python
print(stopwords.fileids())
```

```python
okt = Okt()

example = "고기를 아무렇게나 구울고 하면 안 돼. 고기라고 다 같은 게 아니거든. 예컨대 삼겹살을 구울 때는 중요한 게 있지."
stop_words = "를 아무렇게나 구 우려 고  안 돼 같은 게 구울 때 는"

stop_words = set(stop_words.split(' '))
word_tokens = okt.morphs(example)

result = [word for word in word_tokens if not word in stop_words]

print("불용어 제거 전 :",word_tokens)
print("불용어 제거 후 :",result)
```

```python
import re

# a와 c 사이에 한개의 문자가 포함된 문자열
r = re.compile("a.c")
print(r.search("kkk"))
```

### group, start, end

```python
import re

text = "Hello, world! This is a test."
res = re.search("\\w+", text) # 알파벳 혹은 숫자를 찾음
if res:
    print("매치된 부분:",res.group(0)) # 일치하는 전체 문자열 출력
    print("시작 위치:",res.start()) # 일치하는 부분의 시작 인덱스
    print("종료 위치:",res.end()) # 일치하는 부분의 종료 인덱스
else:
    print("매치되는 부분이 없습니다.")
```

```python
from nltk.tokenize import RegexpTokenizer

text = """Don't be fooled by the dark sounding name, Mr. Jone's Orphanage is as cheery as cheery goes for a pastry shop"""

tokenizer1 = RegexpTokenizer("[\w]+") # 문자 또는 숫자가 한개 이상인 문자열
tokenizer2 = RegexpTokenizer("\s+", gaps=True) # 공백이 하나 이상인 문자열, (gaps = True) -> 패턴과 일치하는 부분을 구분자로 취급

print(tokenizer1.tokenize(text))
print(tokenizer2.tokenize(text))
```

```python
import urllib.request
from soynlp import DoublespaceLineCorpus
from soynlp.word import WordExtractor
```

```python
urllib.request.urlretrieve("https://raw.githubusercontent.com/lovit/soynlp/master/tutorials/2016-10-20.txt", filename="2016-10-20.txt")
```

```python
corpus = DoublespaceLineCorpus("2016-10-20.txt")
len(corpus)
```

```python
i = 0
for document in corpus:
    if len(document) > 0:
        print(document)
        i = i + 1
    print("==========================================")
    if i == 3:
        break
```

```python
word_extractor = WordExtractor()
word_extractor.train(corpus)
word_score_table = word_extractor.extract()
```

```python
word_score_table["반포한"].cohesion_forward # 응집 확률
```

```python
word_score_table["반포한강"].cohesion_forward # 응집 확률
```

```python
word_score_table["반포한강공원에"].cohesion_forward # 응집 확률
```

```python
word_score_table["디스"].right_branching_entropy # 불확실성 수치
```

```python
word_score_table["디스플"].right_branching_entropy # 불확실성이 없음
```

```python
word_score_table["디스플레"].right_branching_entropy
```

```python
from soynlp.tokenizer import LTokenizer

scores = {word:score.cohesion_forward for word, score in word_score_table.items()}
l_tokenizer = LTokenizer(scores=scores)
l_tokenizer.tokenize("국제사회와 우리의 노력들로 범죄를 척결하자", flatten=False)
```

```python
from soynlp.tokenizer import MaxScoreTokenizer

# 띄어쓰기가 되어 있지 않은 문장에서 점수가 높은 글자 시퀀스를 순차적으로 찾아내는 토크나이저
maxscore_tokenizer = MaxScoreTokenizer(scores=scores)
maxscore_tokenizer.tokenize("국제사회와우리의노력들로범죄를척결하자")
```
