```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.impute import SimpleImputer
```

```python
# 데이터 로드
df = pd.read_csv('newbie.csv')

# 열 제거
df = df.drop(columns=['who','Country','Years on Internet'])

# 결측값 처리
imp_mean = SimpleImputer(missing_values=np.nan, strategy='mean')
df['Age'] = imp_mean.fit_transform(df[['Age']])

imp_mode = SimpleImputer(missing_values=np.nan, strategy='most_frequent')
cols = ['Household Income', 'Sexual Preference', 'Marital Status']
df[cols] = imp_mode.fit_transform(df[cols])

# 원핫인코딩
df = pd.get_dummies(df, columns = ['Gender','Household Income','Sexual Preference','Education Attainment','Major Occupation','Marital Status'], dtype='int64')
```

```python
from sklearn.preprocessing import StandardScaler
df2 = df.copy()
scaler = StandardScaler()
df2['Age'] = scaler.fit_transform(df2[['Age']])

X_train, X_test, y_train, y_test = train_test_split(df2.iloc[:,1:], df2.loc[:,"Newbie"], random_state=66, test_size=0.2)
model = LogisticRegression(solver="liblinear")
model.fit(X_train, y_train)
score = model.score(X_test, y_test)
print(f"예측한 결과: {score}")
```

```python
from sklearn.preprocessing import StandardScaler
df2 = df.copy()
scaler = StandardScaler()
df2['Age'] = scaler.fit_transform(df2[['Age']])
```

```python
i=0
max_score = 0
idx = 0
while(i < 100):
    X_train, X_test, y_train, y_test = train_test_split(df2.iloc[:,1:], df2.loc[:,"Newbie"], random_state=i, test_size=0.2)
    model = LogisticRegression(solver="liblinear")
    model.fit(X_train, y_train)
    score = model.score(X_test, y_test)
    print(f"{i}번째 예측한 결과: {score}")
    if max_score < score:
        max_score = score
        idx = i
    i+=1
print(max_score, idx)
```
