```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
```

# 1번

```python
# 타이타닉 데이터 로드
df = pd.read_csv('train.csv')

# 결측치 개수 출력
print(df[['Age','Embarked']].isna().sum())

# 결측치 처리 객체 생성
imp1 = SimpleImputer(missing_values=np.nan, strategy='mean')
imp2 = SimpleImputer(missing_values=np.nan, strategy='most_frequent')

# 결측치 처리 후 저장
df['Age'] = imp1.fit_transform(df[['Age']]).ravel()
df['Embarked'] = imp2.fit_transform(df[['Embarked']]).ravel()

# 결측치 개수 출력
print(df[['Age','Embarked']].isna().sum())
```

# 2번

```python
# 데이터 프레임 복사
df2 = df.copy()

# 레이블 인코더 객체 생성
le = LabelEncoder()

# 레이블 인코딩
sex = le.fit_transform(df2['Sex'])
embarked = le.fit_transform(df2['Embarked'])

# 기존 컬럼 삭제
df2 = df2.drop(['Sex', 'Embarked'],axis=1)

# 새로운 컬럼에 인코딩한 결과 저장
df2['Sex'] = sex
df2['Embarked'] = embarked

# 결과 출력
df2
```

# 3번

```python
# 데이터 프레임 복사
df3 = df.copy()

# 원핫 인코더 객체 생성
enc = OneHotEncoder()

# 적용할 대상
target = ['Sex', 'Embarked']

# 인코딩 결과
result = enc.fit_transform(df3[target])

# 새로운 컬럼 이름
col_names = enc.get_feature_names_out(target)

# 인코딩 결과를 데이터 프레임으로 변환
df_result = pd.DataFrame(result.toarray(), columns=col_names, index=df3.index)

# 기존 컬럼 삭제
df3 = df3.drop(columns=target)

# 인코딩 결과를 df3에 추가
df3 = pd.concat([df3, df_result], axis=1)

# 결과 출력
df3
```
