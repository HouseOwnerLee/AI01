```python
from keras.models import Sequential
from keras.layers import Dense, Dropout, Activation, Flatten
import np_utils # pip install np_utils
from keras.utils import to_categorical
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("./titanic/train.csv")
df
```

```python
df.isna().sum()
```

```python
df = df.fillna(df.mode().iloc[0])
data = df.drop(columns=['Cabin','PassengerId','Name','SibSp','Parch','Ticket','Embarked'])
data
```

```python
data.isna().sum()
```

```python
data["Fare"] = np.log1p(data["Fare"])
data.describe()
```

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
data["Age"] = sc.fit_transform(df[["Age"]])
data.describe()
```

```python
from sklearn.preprocessing import MinMaxScaler

sc = MinMaxScaler()
data["Age"] = sc.fit_transform(df[["Age"]])
data.describe()
```

```python
from sklearn.preprocessing import RobustScaler

sc = RobustScaler()
data["Age"] = sc.fit_transform(df[["Age"]])
data.describe()
```

```python
data.head()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
data["Sex"] = le.fit_transform(data["Sex"])
data["Sex"] = to_categorical(data["Sex"])
data["Pclass"] = to_categorical(data["Pclass"])
```

```python
data.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(data.iloc[:,1:], data.iloc[:,0], test_size=0.2, random_state=42)
X_train.shape, y_train.shape, X_test.shape, y_test.shape
```

```python
from tensorflow.keras.layers import Flatten, Dense, Dropout
from tensorflow.keras import Sequential

model = Sequential()
model.add(Flatten(input_shape=(4,)))
model.add(Dense(4,activation='relu'))
model.add(Dropout(0.2))
model.add(Dense(8,activation='relu'))
model.add(Dropout(0.2))
model.add(Dense(1,activation='sigmoid'))
model.summary()
```

```python
model.compile(optimizer="adam",
              loss="binary_crossentropy",
              metrics=["accuracy"])
```

```python
history = model.fit(X_train, y_train, epochs=200)
```

```python
test_loss, test_acc = model.evaluate(X_test, y_test, verbose=2)
print("\n테스트 정확도 :",test_acc)
```
